```python
from collections import Counter
from pathlib import Path

import cv2
import numpy as np
from PIL import Image

DATASET_ROOT = Path('Dataset/archive')
TRAIN_DIR = DATASET_ROOT / 'asl_alphabet_train' / 'asl_alphabet_train'
TEST_DIR = DATASET_ROOT / 'asl_alphabet_test' / 'asl_alphabet_test'
IMG_SIZE = 64
BATCH_SIZE = 32
VAL_SPLIT = 0.2
SEED = 42
VALID_SUFFIXES = {'.jpg', '.jpeg', '.png', '.bmp'}

def is_valid_image_file(path: Path) -> bool:
    return path.is_file() and path.suffix.lower() in VALID_SUFFIXES and ':' not in path.name

def is_readable_image(path: Path) -> bool:
    try:
        with Image.open(path) as img:
            img.verify()
        return True
    except Exception:
        return False

class_names = sorted([p.name for p in TRAIN_DIR.iterdir() if p.is_dir()])
class_to_index = {name: idx for idx, name in enumerate(class_names)}

all_paths = []
all_labels = []
skipped_files = []

for class_name in class_names:
    class_dir = TRAIN_DIR / class_name
    for img_path in sorted(class_dir.iterdir()):
        if not is_valid_image_file(img_path):
            skipped_files.append(str(img_path))
            continue
        if not is_readable_image(img_path):
            skipped_files.append(str(img_path))
            continue
        all_paths.append(str(img_path))
        all_labels.append(class_to_index[class_name])

rng = np.random.default_rng(SEED)

train_paths, val_paths, train_labels, val_labels = [], [], [], []
for class_name in class_names:
    class_index = class_to_index[class_name]
    class_items = [path for path, label in zip(all_paths, all_labels) if label == class_index]
    class_items = list(rng.permutation(class_items))
    val_count = max(1, int(round(len(class_items) * VAL_SPLIT)))
    val_count = min(val_count, len(class_items) - 1) if len(class_items) > 1 else 1

    val_subset = class_items[:val_count]
    train_subset = class_items[val_count:]

    train_paths.extend(train_subset)
    train_labels.extend([class_index] * len(train_subset))
    val_paths.extend(val_subset)
    val_labels.extend([class_index] * len(val_subset))

train_order = rng.permutation(len(train_paths))
val_order = rng.permutation(len(val_paths))
train_paths = [train_paths[i] for i in train_order]
train_labels = [train_labels[i] for i in train_order]
val_paths = [val_paths[i] for i in val_order]
val_labels = [val_labels[i] for i in val_order]

test_paths = sorted(
    [str(p) for p in TEST_DIR.iterdir() if is_valid_image_file(p) and is_readable_image(p)]
)
test_labels = []
for path_str in test_paths:
    stem = Path(path_str).stem
    class_name = stem.replace('_test', '')
    test_labels.append(class_to_index.get(class_name))

num_classes = len(class_names)

def load_image(path):
    image = cv2.imread(str(path))
    if image is None:
        raise ValueError(f'Could not read image: {path}')
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_AREA)
    return image.astype(np.float32) / 255.0

train_counts = Counter(train_labels)
val_counts = Counter(val_labels)

print(f'Dataset root: {DATASET_ROOT.resolve()}')
print(f'Classes ({num_classes}): {class_names}')
print(f'Total readable training images: {len(all_paths)}')
print(f'Train samples: {len(train_paths)}, Val samples: {len(val_paths)}')
print(f'Test samples: {len(test_paths)}')
print(f'Skipped invalid/unreadable files: {len(skipped_files)}')
if skipped_files:
    print('Example skipped file:', skipped_files[0])
print('Train distribution:', {class_names[idx]: train_counts[idx] for idx in range(num_classes)})
print('Val distribution:', {class_names[idx]: val_counts[idx] for idx in range(num_classes)})
```

```python
import math

def batch_generator(paths, labels, batch_size, shuffle=True, rng=None, drop_remainder=False):
    n = len(paths)
    indices = np.arange(n)
    if shuffle:
        if rng is None:
            indices = np.random.permutation(indices)
        else:
            indices = rng.permutation(indices)

    for start in range(0, n, batch_size):
        batch_idx = indices[start : start + batch_size]
        if len(batch_idx) < batch_size and drop_remainder:
            continue

        batch_x, batch_y = [], []
        for i in batch_idx:
            try:
                batch_x.append(load_image(paths[i]))
                batch_y.append(labels[i])
            except Exception as err:
                print(f"Skipping unreadable image: {paths[i]} ({err})")

        if not batch_x:
            continue

        batch_x = np.stack(batch_x, axis=0)
        batch_y = np.eye(num_classes, dtype=np.float32)[np.array(batch_y, dtype=np.int32)]
        yield batch_x, batch_y

def infinite_batch_generator(paths, labels, batch_size, shuffle=True, rng=None, drop_remainder=False):
    while True:
        yield from batch_generator(
            paths,
            labels,
            batch_size,
            shuffle=shuffle,
            rng=rng,
            drop_remainder=drop_remainder,
        )

print(f"Classes: {num_classes}")
print(f"Train samples: {len(train_paths)}, Val samples: {len(val_paths)}")
print(f"Image shape: (IMG_SIZE, IMG_SIZE, 3) = ({IMG_SIZE}, {IMG_SIZE}, 3)")
```

```python
import keras
model = keras.Sequential(
    [
        keras.Input(shape=(IMG_SIZE, IMG_SIZE, 3)),
        keras.layers.Conv2D(32, kernel_size=(3, 3), activation="relu"),
        keras.layers.MaxPooling2D(pool_size=(2, 2)),
        keras.layers.Conv2D(64, kernel_size=(3, 3), activation="relu"),
        keras.layers.MaxPooling2D(pool_size=(2, 2)),
        keras.layers.Flatten(),
        keras.layers.Dropout(0.5, seed=SEED),
        keras.layers.Dense(num_classes, activation="softmax"),
    ]
)


model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"],
)
model.summary()
```

```python
import cv2
import matplotlib.pyplot as plt

#read Dataset/archive/asl_alphabet_test/asl_alphabet_test/A_test.jpg
img = cv2.imread('Dataset/archive/asl_alphabet_test/asl_alphabet_test/A_test.jpg')
plt.imshow(img)
plt.show()

```

```python
# Train model with explicit generator steps
train_steps = len(train_paths) // BATCH_SIZE
val_steps = math.ceil(len(val_paths) / BATCH_SIZE)

train_generator = infinite_batch_generator(
    train_paths,
    train_labels,
    BATCH_SIZE,
    shuffle=True,
    rng=np.random.default_rng(SEED),
    drop_remainder=True,
)
val_generator = infinite_batch_generator(
    val_paths,
    val_labels,
    BATCH_SIZE,
    shuffle=False,
    drop_remainder=False,
)

history = model.fit(
    train_generator,
    steps_per_epoch=train_steps,
    validation_data=val_generator,
    validation_steps=val_steps,
    epochs=10,
)
```

```python
# Evaluate on validation set (JAX model)
y_pred_list, y_true_list = [], []
for x, y in batch_generator(val_paths, val_labels, BATCH_SIZE, shuffle=False):
    pred = model.predict(np.array(x))
    y_pred_list.append(np.array(np.argmax(pred, axis=-1)))
    y_true_list.append(np.argmax(y, axis=-1))
y_pred = np.concatenate(y_pred_list)
y_true = np.concatenate(y_true_list)

acc = np.mean(y_pred == y_true)

# Weighted F1 with NumPy
f1_sum = 0.0
for c in range(num_classes):
    tp = np.sum((y_true == c) & (y_pred == c))
    fp = np.sum((y_true != c) & (y_pred == c))
    fn = np.sum((y_true == c) & (y_pred != c))
    p = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    r = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1_c = 2 * p * r / (p + r) if (p + r) > 0 else 0.0
    f1_sum += f1_c * np.sum(y_true == c)
f1 = f1_sum / len(y_true)

print(f"Validation Accuracy: {acc:.4f}")
print(f"Validation F1 Score (weighted): {f1:.4f}")
```

```python
# Plot training history
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history["accuracy"], label="Train")
axes[0].plot(history["val_accuracy"], label="Val")
axes[0].set_title("Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].legend()

axes[1].plot(history["loss"], label="Train")
axes[1].plot(history["val_loss"], label="Val")
axes[1].set_title("Loss")
axes[1].set_xlabel("Epoch")
axes[1].legend()

plt.tight_layout()
plt.show()
```
