# Predict One Video

Load one saved model run, load one video, and show the prediction.

```python
from pathlib import Path

# Change these two paths for your test.
MODEL_PATH = Path("artifacts/experiments/si/transformer/hand_body/20260618T155746Z")
VIDEO_PATH = Path("address.mp4")
TOP_K = 5

MODEL_PATH, VIDEO_PATH
```

```python
import os

os.environ.setdefault("KERAS_BACKEND", "torch")

from IPython.display import Video, display

from training.testing_pipeline import load_run_config, predict_video

run_dir, config = load_run_config(MODEL_PATH)
print("Run directory:", run_dir)
print("Model:", config["model"])
print("Modality:", config["modality"])
print("Input shape:", config["input_shape"])
print("Classes:", config["num_classes"])
```

```python
if not VIDEO_PATH.exists():
    raise FileNotFoundError(f"Video not found: {VIDEO_PATH}")

display(Video(str(VIDEO_PATH), embed=True, width=640))
```

```python
result = predict_video(
    video_path=VIDEO_PATH,
    model_path=MODEL_PATH,
    top_k=TOP_K,
)

top_prediction = result["top_predictions"][0]
print(f"Prediction: {top_prediction['label']} ({top_prediction['confidence']:.4f})")
print()
print("Top predictions:")
for item in result["top_predictions"]:
    print(f"{item['rank']:>2}. {item['label']} [{item['index']}] {item['confidence']:.4f}")
```
